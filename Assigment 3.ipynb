```python
num1 = float(input("Enter first number: "))
operator = input("Enter operator (+, -, *, /, **): ")
num2 = float(input("Enter second number: "))

if operator == "+":
    print("Result:", num1 + num2)
elif operator == "-":
    print("Result:", num1 - num2)
elif operator == "*":
    print("Result:", num1 * num2)
elif operator == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Error: Division by zero")
elif operator == "**":
    print("Result:", num1 ** num2)
else:
    print("Invalid operator")
```

```python
my_list = [10, "hello", 5.5, "world"]

has_number = False
for item in my_list:
    if isinstance(item, (int, float)):
        has_number = True
        break

if has_number:
    print("List contains a numeric value")
else:
    print("No numeric value in the list")
```

```python
my_dict = {"name": "Salman", "age": 22}

# Add new key
my_dict["city"] = "Karachi"

print("Updated dictionary:", my_dict)
```

```python
my_dict = {"a": 10, "b": 20, "c": "hello", "d": 30}

total = 0
for value in my_dict.values():
    if isinstance(value, (int, float)):
        total += value

print("Sum of numeric values:", total)
```

```python
my_list = [1, 2, 3, 2, 4, 5, 1, 6]

duplicates = []
for item in my_list:
    if my_list.count(item) > 1 and item not in duplicates:
        duplicates.append(item)

print("Duplicate values:", duplicates)
```
